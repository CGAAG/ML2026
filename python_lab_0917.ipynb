```python
# 1부터 100까지의 숫자를 생성합니다.
numbers = range(1, 101)
# 생성된 숫자들의 합계를 계산합니다.
sum_numbers = sum(numbers)
# 생성된 숫자들의 개수를 계산합니다.
count_numbers = len(numbers)
# 합계를 개수로 나누어 평균을 계산합니다.
average = sum_numbers / count_numbers
# 계산된 평균을 출력합니다.
print(f"1부터 100까지 숫자의 평균: {average}")
```

### 사용된 함수 설명

*   **`range(start, stop)`**: `start`부터 `stop - 1`까지의 정수 시퀀스를 생성하는 함수입니다. 이 코드에서는 `range(1, 101)`을 사용하여 1부터 100까지의 숫자를 만듭니다.

*   **`sum(iterable)`**: 반복 가능한(iterable) 객체(예: 리스트, 튜플, `range` 객체)에 있는 모든 항목의 합계를 계산하는 함수입니다.

*   **`len(s)`**: 객체(문자열, 튜플, 리스트, 딕셔너리 등)의 길이(항목 수)를 반환하는 함수입니다. 이 코드에서는 `range` 객체에 포함된 숫자의 개수를 세는 데 사용됩니다.

```python
# 정수(Integer) 타입
a = 123
print(type(a))  # 결과: <class 'int'>

# 실수(Float) 타입
a = 123.456
print(type(a))  # 결과: <class 'float'>

# 문자열(String) 타입
a = "Hello World!"
print(type(a))  # 결과: <class 'str'>

# 불리언(Boolean - 참/거짓) 타입
a = True
print(type(a))  # 결과: <class 'bool'>

# 리스트(List - 배열) 타입
a = [1, 2, 3]
print(type(a))  # 결과: <class 'list'>
```

```python
# 불리언(Boolean) 데이터의 산술 연산 (True는 1, False는 0으로 처리됨)
a = True; b = False
print(a + b)  # 결과: 1 (1 + 0의 결과)

# 변수에 정수를 할당하고 덧셈(산술 연산) 수행
a = 3; b = 4
c = a + b
print(c)  # 결과: 7

# 비교 연산자 사용 (a가 b보다 작은지 비교)
d = a < b
print(d)  # 결과: True (3은 4보다 작으므로 '참')

# 논리 연산자(and) 사용 (양쪽의 조건이 모두 참인지 확인)
e = 3 < 4 and 4 < 5
print(e)  # 결과: True (3 < 4도 참이고, 4 < 5도 참이므로 최종 결과는 '참')
```

```python
# 1. 정수(Integer)형: 소수점이 없는 숫자를 저장합니다.
a = 123
print(type(a))  # 출력 결과: <class 'int'>

# 2. 실수(Float)형: 소수점이 있는 숫자를 저장합니다.
a = 123.456
print(type(a))  # 출력 결과: <class 'float'>

# 3. 문자열(String)형: 텍스트 데이터를 저장하며 큰따옴표("")나 작은따옴표('')로 감쌉니다.
a = "Hello World!"
print(type(a))  # 출력 결과: <class 'str'>

# 4. 불리언(Boolean)형: 참(True) 또는 거짓(False)이라는 논리값을 저장합니다.
a = True
print(type(a))  # 출력 결과: <class 'bool'>

# 5. 리스트(List)형: 여러 개의 값을 대괄호 [] 안에 순서대로 나열하여 모아두는 배열입니다.
a = [1, 2, 3]
print(type(a))  # 출력 결과: <class 'list'>
```

핵심 함수 설명
type(값): 괄호 안에 들어간 변수나 데이터의 자료형(타입)을 확인하여 알려주는 파이썬 내장 함수입니다. 파이썬은 변수에 값을 넣을 때 자동으로 자료형을 결정하므로, 이 함수를 통해 현재 변수가 어떤 형태의 데이터를 담고 있는지 확인할 수 있습니다.

print(값): 괄호 안의 최종 결과를 화면(콘솔)에 텍스트로 출력해 주는 함수입니다. 여기서는 type(a)가 판별해낸 자료형 결과를 눈으로 볼 수 있게 출력하는 역할을 합니다.
